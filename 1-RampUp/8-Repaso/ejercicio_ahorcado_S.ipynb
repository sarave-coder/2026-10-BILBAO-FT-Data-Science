## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".

````python
#mostrar rayas bajas

import random

lista_palabras= ["cerveza", "comer", "caminata", "hotel", "sonrisa", "playa", "amanecer"]
palabra2 = random.choice(lista_palabras)
rayas = []
for i in palabra2:
    rayas.append("_")

print(rayas)
print(palabra2)
````

````python
n_vidas = 6

import time

while n_vidas >0:

    letra_usuario=input("dime una letra").lower()

    if letra_usuario in palabra2:
         print("Has acertado")

    else:
        n_vidas = n_vidas - 1
        print("Pierdes una vida")

    for indice, letra in enumerate(palabra2):

        if letra_usuario == letra:
            rayas[indice] = letra_usuario

    print(" ".join(rayas))
   
    time.sleep(0.1)

if n_vidas == 0:
    print("Fin del juego!")
````
