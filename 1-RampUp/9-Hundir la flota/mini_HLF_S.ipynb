```python
#Para introducir en consola
def imprimirTablero(tablero):
    for fila in tablero:
        for columna in fila:
            
            print(f"[{columna}]", end=" ")
        print(" ")
```

```python
# función disparar para utilizar en While para ambos jugador y maquina

def disparar(tablero, input_i, input_j):
   
    if tablero[input_i][input_j] == "B":
        tablero[input_i][input_j] = "X"
        print (f"tocado en {input_i},{input_j}")
        imprimirTablero(tablero)
        return True

    else:
        tablero[input_i][input_j] = "O"
        print (f"agua en {input_i},{input_j}")
        imprimirTablero(tablero)
        return False
    
```
