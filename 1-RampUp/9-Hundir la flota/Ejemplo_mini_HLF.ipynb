## Mini Hundir la Flota
#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
* Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i y otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
'''
Crear un Tablero como una lista de listas de 10x10 que contenga caracteres espacio: " "
'''
import pprint
dimension_tablero = 10
tablero = [[" " for i in range(dimension_tablero)] for j in range(dimension_tablero)]
pprint.pprint(tablero)
```

```python
'''
Insertar dos barcos sustituyendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
'''

flota=[{"Nombre": "Barco_1", "Eslora": 4, "x": 1, "y": 0, "Orientación": "h", "Vidas": 4},
       {"Nombre": "Barco_2", "Eslora": 3, "x": 3, "y": 3, "Orientación": "v", "Vidas": 3}]

for barco in flota:
    x = barco["x"]
    y = barco["y"]
    contador = 0
    while contador < barco["Eslora"]:
        if barco["Orientación"] == "h":
            tablero[x][y] = "B"
            y = y + 1
            contador = contador + 1
        else:
            tablero[x][y] = "B"
            x = x + 1
            contador = contador + 1

pprint.pprint(tablero)
```

```python
'''
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haya una en "B")
        * Si hay una "B":
            * Cambiar el contenido de esa posición por una "x"
            * Imprimir por pantalla el texto "Tocado en posición `i,j`"
        * Si hay un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el texto "Agua"
    * Imprimir por pantalla el Tablero
'''

def efectuando_un_disparo():
    try:
        i = int(input(f"Introduzca la fila (0-9)"))
        j = int(input(f"Introduzca la columna (0-9)"))

    except ValueError:          
        print("El valor introducido no es válido.")         #Compruebo que el usuario está metiendo un número
        return
    
    if not (0 <= i <= 9 and 0 <= j <= 9):                   
        print("La coordenada elegida está fuera del tablero.")          #Compruebo que i y j quedan dentro del tablero
        return
    
    disparo = tablero[i][j]

    if disparo == "B":
        tablero[i][j] = "x"
        print (f"Tocado en posición ({i},{j})")

    elif disparo == " ":
        tablero[i][j] = "o"
        print(f"Agua en la posición ({i},{j})")

    else:
        print(f"Ya habías disparado en ({i},{j}), elige otra coordenada")

    pprint.pprint(tablero)
```

```python
'''
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una para i y otra para j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
'''
def solicitar_coordenada():
    while True:
        try:
            i = int(input(f"Introduzca la fila (0-9)"))
            j = int(input(f"Introduzca la columna (0-9)"))
            if 0 <= i <= 9 and 0 <= j <= 9:
                return i,j 
            else:
                print(f"La coordenada ({i},{j}) está fuera del tablero.")
        except ValueError:
            print("El valor introducido no es válido.")
            
```

```python
solicitar_coordenada()
```

```python
def disparar(i, j):

    disparo = tablero[i][j]
    
    if disparo == "B":
        tablero[i][j] = "x"
        print (f"Tocado en posición ({i},{j})")
        return True
    
    elif disparo == " ":
        tablero[i][j] = "o"
        print(f"Agua en la posición ({i},{j})")
        return False
    
    else:
        print(f"Ya habías disparado en ({i},{j}), elige otra coordenada")
        return False
    
```

```python
while True:

    i,j = solicitar_coordenada()

    acierto = disparar(i, j)

    pprint.pprint(tablero)

    if not acierto:
        print("Fin del turno")
        break
    else:
        print (f"Vuelves a disparar")
        print()

```
