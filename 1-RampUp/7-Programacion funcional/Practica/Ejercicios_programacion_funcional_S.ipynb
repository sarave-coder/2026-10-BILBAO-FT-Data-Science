### 2. Filtrar números pares de una lista:

Emplea filter para obtener sólo los números pares de la lista [1, 2, 3, 4, 5, 6].

Solución esperada: [2, 4, 6]

```python
lista = [1,2,3,4,5,6]

filtrado = filter(lambda n: n %2 ==0, lista)

print(list(filtrado))
```

### 3. Convertir grados Celsius a Fahrenheit:

Usa map y una función lambda para convertir una lista de temperaturas en grados Celsius [0, 10, 20, 30] a Fahrenheit.

Solución esperada: [32.0, 50.0, 68.0, 86.0]

```python
celsius = [0,10,20,30]

grados_f = list(map(lambda c: c * 9 / 5 + 32, celsius))
print(grados_f)

# función 9/5 + 32 es para pasar de celsius a far.
```

### 4. Sumar los cuadrados de los números pares de una lista:

Combina filter, map y reduce (importa reduce desde functools) para sumar los cuadrados de los números pares en [1, 2, 3, 4, 5, 6].

Solución esperada: 56

```python
from functools import reduce
```

```python
number = [1,2,3,4,5,6]
filtrado = filter(lambda n: n %2 ==0, number)
cuadrados = map(lambda n: n ** 2, filtrado)
resultado = reduce(lambda x,y: x+y, list(cuadrados))
print(resultado)
```

```python
number = [1,2,3,4,5,6]

# Filtrar solo los números pares

filtrado = list(filter(lambda n: n %2 ==0, number))

print(filtrado)

# Elevar al cuadrado cada número par

cuadrados = list(map(lambda n: n ** 2, filtrado))

print(type (cuadrados))
print(cuadrados)

# Sumar todos los cuadrados

resultado = reduce(lambda x,y: x+y, list(cuadrados))
print(resultado)
```

### 5. Encadenar map, filter, y reduce para resolver un problema complejo:

Dada una lista de strings que representan números enteros y "NA" (por ejemplo, ["2", "5", "NA", "9", "10"]), convierte los números a enteros, filtra los "NA", suma el cuadrado de los números que son mayores que 5 y menores que 10.

Solución esperada: Un único valor que es la suma de los cuadrados de los números válidos.

```python
lista_str = ["2","5","NA","9","10"]

#filtar NA
filtarNA = list(filter(lambda n: n != "NA", lista_str))

print(filtarNA)

#convertir a entero
n_entero = list(map(lambda n: int(n), filtarNA))

print(n_entero)


# sumar cuadrado de los numeros mayores 5 y menores 10

ordenar = list(filter(lambda n: 5 < n < 11, n_entero)) 

print(ordenar)

cuadrado = list(map(lambda n : n**2, ordenar))

print(cuadrado)

#resultado

resultado = reduce(lambda x,y: x+y, list(cuadrado))
print(resultado)
```
