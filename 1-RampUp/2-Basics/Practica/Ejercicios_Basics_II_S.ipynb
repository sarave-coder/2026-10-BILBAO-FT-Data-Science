## Ejercicio 5
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
input1 = float(input("Escribe el primer dato"))
input2 = float(input("Escribe el segundo dato"))
if input1 == input2:
    print(f"Los datos {input1} y {input2} son iguales")

else:
    print(f"Los datos {input1} y {input2} no son iguales")
````

## Ejercicio 6
Mismo programa que en 5, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
D = float(input("Escribe el primer dato"))
E = float(input("Escribe el segundo dato"))
F = float(input("Escribe el tercer dato"))
if D == E and D == F:
    print(f"Todos los datos son iguales")

if D == E or D == F or E == F:
    print(f"Al menos dos datos son iguales")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
G = float(input("Escribe primer dato"))
H = float(input("Escribe segundo dato"))
suma = G + H
if suma < 10:
    print(f"La suma es inferior a 10")

elif suma > 10:
    print(f"La suma es superior a 10")

else:
    print(f"La suma es igual a 10")
````
