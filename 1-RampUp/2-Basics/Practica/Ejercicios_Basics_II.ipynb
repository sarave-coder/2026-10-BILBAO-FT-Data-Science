## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

$$radianes = grados*(pi/180)$$

````python
import math
````

````python
grados = float(input("Introduce los grados que desea conertir: "))
radianes = grados * (math.pi / 180)
print(f"{grados} grados equivalen a {radianes} radianes")
````

## Ejercicio 5
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
Input_1 = int(input("Introduce un número: "))
Input_2 = int(input("Introduce otro número: "))

if Input_1 > Input_2:
    print(f"El número {Input_1} es mayor que {Input_2}")

elif Input_1 < Input_2:
    print(f"El número {Input_1} es menor que {Input_2}")

else:
    print(f"El número {Input_1} es igual que {Input_2}")
    
````
