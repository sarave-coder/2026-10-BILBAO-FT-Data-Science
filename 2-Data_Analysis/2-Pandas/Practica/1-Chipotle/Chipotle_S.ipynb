```python
# Hay que calcular el precio unitario

chipo['item_price'] = pd.to_numeric(chipo['item_price'].astype(str).str.replace('$','', regex=False), errors='coerce') # quitar los dolares y convirtirlo a str

chipo
```

```python
# Pedido más caro
print(chipo.loc[chipo["item_price"].idxmax(), ["item_name", "item_price", "quantity"]])

#sacar el producto con el precio más alto, el producto, precio y cantidad
```

```python
# calcular el precio unitario para saber el producto más caro por unidad
# creación de la variable 'unit_price'

chipo['unit_price']= chipo ['item_price']/ chipo ['quantity']
chipo.head(10)
```

```python
print(chipo.loc[chipo['unit_price'].idxmax(),['item_name','item_price','quantity', 'unit_price']])
```

```python
#más de un producto
productos_caros = chipo[chipo['unit_price'] == chipo['unit_price'].max()]
print(productos_caros['item_name'].unique())
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
#cantidad_pedidos = chipo.loc[cantidad_productos > 10, "order_id"].nunique()

#print(cantidad_pedidos)

# chipo.loc[cantidad_productos > 10, "order_id"] 
# loc permite seleccionar filas y columnas de un DataFrame.
# cantidad_productos > 10: selecciona las filas que cumplen la condición.
# "order_id": selecciona únicamente la columna que contiene los identificadores de los pedidos.
# nunique: Cuenta los valores únicos, es decir, los identificadores de pedido diferentes, sin contar repetidos.
```

```python
print(chipo.loc[chipo["unit_price"] > 10, "order_id"].nunique())
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
cantidad_productos = chipo.groupby(chipo["order_id"])["item_price"].sum()
print(len(cantidad_productos > 10))

#Cuenta los pedidos que contienen más de 10 unidades de productos, independientemente de su precio.
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
print(chipo.loc[(chipo['item_price']>10) & (chipo['quantity']>1),'order_id'].nunique())
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
idx_implicito = chipo['item_price'].argmax() #argmax indice implicito, no puedes cambiar. Pides la fila
idx_explicito = chipo['item_price'].idxmax() # pides el indice explicito 

print(chipo.loc[idx_explicito, 'quantity']) 

#print(chipo.iloc[idx_implicito, 'quantity']) # posición de la fila en ese momento

print(chipo.loc[idx_explicito])
print(chipo.iloc[idx_implicito])
```
