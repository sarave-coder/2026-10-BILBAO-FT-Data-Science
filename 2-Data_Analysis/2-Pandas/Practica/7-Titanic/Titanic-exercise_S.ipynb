20. Calculate the ratio of missing values at the `boat` column. 

```python
titanic['boat'] = np.nan
titanic['boat'] = titanic['boat'].astype('object')

#titanic.loc[:485, 'boat'] = 'some_value'  # 486 non-missing values
```

```python
# Total rows
total_rows = len(titanic)

# Non-missing values in 'boat'
non_missing = titanic['boat'].count()
```

33. Using the function [`catplot`](https://seaborn.pydata.org/generated/seaborn.catplot.html) from the library `seaborn`, find out if the hypothesis _"Women are more likely to survive shipwrecks"_ is true or not.

You should get something like this:

![catplot](./img/catplotgen.png)

```python
# Crear: categorical plot: survival by gender
sns.catplot(
    x='gender',
    y='survived',
    data=titanic,
    kind='bar',            # bar plot for average values
    palette=['skyblue', 'orange'],
    height=5,
    aspect=1
)

plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')
plt.ylim(0, 1)  # survival rate between 0 and 1
plt.show()

# Calculate survival rates numerically
survival_rates = titanic.groupby('gender')['survived'].mean()
print("\nSurvival rates by gender:")
print(survival_rates)
```

34. Using [`kdeplot`]("https://seaborn.pydata.org/generated/seaborn.kdeplot.html") from `seaborn` represent those who not survived distributed by age.<>

Hint: First you should "filter" the `titanic` dataset where the column "survived" is 0, indexing the column `"age"` only.

Arguments you should pass to the function: 

    - color = "red"
    - label = "Not Survived"
    - shade = True
    
You should get something like this: 

![kdeplot](./img/kdeplotsur.png)

```python
# filtrado de pasajeros por edad que no han sobrevivido 
ages_not_survived = titanic.loc[titanic['survived'] == 0, 'age'].dropna()

# crear KDE plot
plt.figure(figsize=(8, 5))
sns.kdeplot(
    ages_not_survived,
    color="red",
    label="Not Survived",
    shade=True
)

plt.legend('Not Survive')

plt.show()
```
