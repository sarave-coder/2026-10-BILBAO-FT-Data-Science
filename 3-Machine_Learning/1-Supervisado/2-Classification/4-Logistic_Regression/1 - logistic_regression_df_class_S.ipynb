## Creamos el modelo

```python
X = np.array(dataframe.drop(columns=['clase'])) # drop de y, que es el target y lo separas de X
y = np.array(dataframe['clase'])
```

```python
X = dataframe.drop("clase",axis=1).values # axis, para quitar la columna. Es lo mismo que np.array
y = dataframe["clase"].values
```

```python
X
```

```python
y # clase
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000) # parámetro de la regresión logísitca. cuantas veces va a dar vueltas para buscar una solución
model.fit(X, y)

# con pocas filas y columnas, 1000 está bien. Si es un problema más complejo se utilizará otro método. Por lo que 1000 es ok
```

```python
predictions = model.predict(X)
print(predictions)
```

```python
predicions_proba = model.predict_proba(X) # probabilidades, en el caso 1 tiene x probabilidades de ser un windows, mac...(clase)
print(np.round(np.array(predicions_proba), 2))
#predicions_proba
# columna 1(windows), 2(linux)...
```

```python
model.score(X, y)

# modelo score de clasificación te da el Acuraccy; cuántos ha acertado de todos los que ha predicho. Ha acertado un 77%
```

```python
model.classes_
```

# Validación del Modelo

```python
validation_size = 0.20
seed = 7
X_train, X_test, Y_train, Y_test = model_selection.train_test_split(X,
                                                                    y,
                                                                    test_size=validation_size,
                                                                    random_state=seed)
```

```python
name='Logistic Regression'
kfold = model_selection.KFold(n_splits=10) #Parte los datos en 10 trozos para usar validación cruzada / cross validation
cv_results = model_selection.cross_val_score(model, X_train, Y_train, cv=kfold, scoring='accuracy')

msg = "%s: %f (%f)" % (name, cv_results.mean(), cv_results.std())
print(cv_results)
print(msg)

# métricas del modelo de las 10 maneras distintas, cómo de bueno o malo es el modelo, validación cruzada
# (logistic regression), media y desviación estándar
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, Y_train)
```

## Resultados

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
```

```python
predictions = model.predict(X_test) # comienzo a testear 
print(accuracy_score(Y_test, predictions))
predictions

# como había pocos valores el acuracy ha mejorado
```

```python
acierto = accuracy_score(Y_test, predictions)

error = 1 - acierto
print("Acierto:", round(acierto*100, 2), "%")
print("Error:", round(error*100, 2), "%")
```

```python
c_matrix = confusion_matrix(Y_test, predictions)
```

```python
print(c_matrix)

# columna 1, windows...hacia abajo predichos y hacia la derecha reales
```

```python
import seaborn as sns
sns.heatmap(c_matrix, annot=True);
```

```python
sns.heatmap(confusion_matrix(Y_test, predictions, normalize='true'), annot=True, 
            fmt='.2%', cmap='Blues');
```

```python
from sklearn.metrics import multilabel_confusion_matrix, precision_score, recall_score, f1_score

multilabel_confusion_matrix(Y_test, predictions)
```
